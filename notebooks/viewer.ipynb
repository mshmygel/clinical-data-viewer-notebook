```python
import pandas as pd
import pyreadstat
from pathlib import Path

# Supported extensions
SUPPORTED_EXTENSIONS = [".sas7bdat", ".xpt", ".csv", ".xlsx"]

# Data folder
DATA_DIR = Path("../data")

def list_datasets():
    return [f.name for f in DATA_DIR.iterdir() if f.suffix.lower() in SUPPORTED_EXTENSIONS]

def read_dataset(file_name):
    path = DATA_DIR / file_name
    ext = path.suffix.lower()

    if ext == ".sas7bdat":
        df, _ = pyreadstat.read_sas7bdat(path)
    elif ext == ".xpt":
        df, _ = pyreadstat.read_xport(path)
    elif ext == ".csv":
        df = pd.read_csv(path)
    elif ext in [".xls", ".xlsx"]:
        df = pd.read_excel(path)
    else:
        raise ValueError(f"Unsupported file type: {ext}")

    return df
```

```python
# List available datasets
print("Available files:")
for f in list_datasets():
    print(" -", f)

# Choose a file
file_name = input("\nEnter file name to load: ")

# Load and preview
df = read_dataset(file_name)

print("\nLoaded:", file_name)
print("Shape:", df.shape)

# Show head
display(df.head())

# Show info
print("\nDataFrame Info:")
df.info()

# Describe numerics
print("\nStatistical Summary:")
display(df.describe(include='all'))
```

```python
# Show missing values per column (critical for clinical data completeness)
null_counts = df.isnull().sum()
null_pct = (null_counts / len(df)) * 100

null_summary = pd.DataFrame({
    "Null Count": null_counts,
    "Percent Null (%)": null_pct.round(2)
}).sort_values(by="Percent Null (%)", ascending=False)

print("Missing value summary:")
display(null_summary)
```

```python
# Save current DataFrame to file
output_file = input("Enter filename to save (with .xlsx or .csv): ")

if output_file.endswith(".xlsx"):
    df.to_excel(output_file, index=False)
    print("Saved to", output_file)
elif output_file.endswith(".csv"):
    df.to_csv(output_file, index=False)
    print("Saved to", output_file)
else:
    print("Unsupported file format. Use .xlsx or .csv.")
```
